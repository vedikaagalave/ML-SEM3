```python
#flatten() vs ravel()
arr = np.arange(12).reshape(4,3)
print(arr)
arr1 = arr.ravel() # shallow copy if u change here it will also change in flatten one
arr1[0] = 99
arr1[2] = 98
arr2 = arr.flatten() # deep copy if u change here only change here
arr2[1] = 100

print("ravel arr",arr1)
print("flatten arr",arr2)
```

```python
# Create a blank mark sheet for 6 students and 4 subjects, every entry = 0
sheet = np.zeros((6,4), dtype = int)
print(sheet)
# Same size where every student is temporarily given 35
sheet1 = np.full((6,4),35)
print(sheet1)
# Alternate roll numbers from 101 to 130
sheet2 = np.arange(101,131,2)
print(sheet2)
# Split 3 hours into 7 equally spaced time markers
sheet3 = np.linspace(0,3,7)
print(sheet3)
# 5x5 matrix with 10 on diagonal and 0 elsewhere
sheet4 = np.eye(5,5)*10
print(sheet4)
# 4x4 matrix with 100 on diagonal and 35 elsewhere
sheet5 = np.full((4,4),35)
np.fill_diagonal(sheet5,100)
print(sheet5)
```
